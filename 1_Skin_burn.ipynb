```python
import os
import cv2
import numpy as np
from sklearn.model_selection import train_test_split
from keras.applications import MobileNet
from keras.models import Sequential
from keras.layers import GlobalAveragePooling2D, Dense
from matplotlib import pyplot as plt
from tensorflow.keras.optimizers import Adam
```

```python
image_folder = '/content/drive/MyDrive/microsoft_hackerground/data/'
class_folders = ['/content/scaled_1st degree burn',
                 '/content/scaled_2nd degree burn',
                 '/content/scaled_3nd degree burn']
classes = len(class_folders)
```

```python
images = []
labels = []
```

```python
for i, folder in enumerate(class_folders):
    folder_path = os.path.join(image_folder, folder)
    image_names = os.listdir(folder_path)
    for image_name in image_names:
        image = cv2.imread(os.path.join(folder_path, image_name))
        image = cv2.resize(image, (224, 224))
        images.append(image)
        labels.append(i)
```

```python
images = np.array(images)
labels = np.array(labels)
```

```python
train_images, test_images, train_labels, test_labels = train_test_split(images, labels, test_size=0.2, random_state=42)
```

```python
base_model = MobileNet(weights='imagenet', include_top=False, input_shape=(224, 224, 3))
```

```python
model = Sequential()
model.add(base_model)
model.add(GlobalAveragePooling2D())
model.add(Dense(64, activation='relu'))
model.add(Dense(128, activation='selu'))
model.add(Dense(classes, activation='softmax'))
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

history = model.fit(train_images, train_labels, epochs=15, batch_size=32, validation_data=(test_images, test_labels))
accuracy = history.history['accuracy']
val_accuracy = history.history['val_accuracy']
loss = history.history['loss']
val_loss = history.history['val_loss']
epochs = range(len(accuracy))

plt.plot(epochs, accuracy, 'bo', label='Training accuracy')
plt.plot(epochs, val_accuracy, 'b', label='Validation accuracy')
plt.title('Training and validation accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()
plt.show()

plt.plot(epochs, loss, 'bo', label='Training loss')
plt.plot(epochs, val_loss, 'b', label='Validation loss')
plt.title('Training and validation loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.show()
```

```python
validation_data = (test_images, test_labels)
evaluation = model.evaluate(test_images, test_labels)
print("Evaluation loss:", evaluation[0])
print("Evaluation accuracy:", evaluation[1])
```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing import image
from tensorflow.keras.models import load_model
import matplotlib.pyplot as plt
import cv2

# Save the trained model to a directory
model.save('Burn_model.keras')

# Download the saved model directory to your local machine
!zip -r Burn_model.zip Burn_model.keras
from google.colab import files
files.download('Burn_model.zip')

# Load the trained model from the directory
model = load_model('/content/Burn_model.keras')

# Load and preprocess the image
def preprocess_input_image(img_path):
    img = image.load_img(img_path, target_size=(224, 224))
    img_array = image.img_to_array(img)
    img_array = np.expand_dims(img_array, axis=0)
    img_array = tf.keras.applications.mobilenet.preprocess_input(img_array)
    return img_array

# Function to generate Grad-CAM heatmap
def make_gradcam_heatmap(model, img_array, last_conv_layer_name, pred_index=None):
    grad_model = tf.keras.models.Model([model.inputs], [model.get_layer(last_conv_layer_name).output, model.output])

    with tf.GradientTape() as tape:
        last_conv_layer_output, preds = grad_model(img_array)
        if pred_index is None:
            pred_index = tf.argmax(preds[0])
        class_channel = preds[:, pred_index]

    grads = tape.gradient(class_channel, last_conv_layer_output)
    pooled_grads = tf.reduce_mean(grads, axis=(0, 1, 2))
    last_conv_layer_output = last_conv_layer_output[0]
    heatmap = last_conv_layer_output @ pooled_grads[..., tf.newaxis]
    heatmap = tf.squeeze(heatmap)

    heatmap = tf.maximum(heatmap, 0) / tf.math.reduce_max(heatmap)
    return heatmap.numpy()

# Load and preprocess the image
img_path = '/content/scaled_1st degree burn/sample_image.jpg'
img_array = preprocess_input_image(img_path)

# Generate Grad-CAM heatmap
heatmap = make_gradcam_heatmap(model, img_array, last_conv_layer_name='conv_pw_13_relu')

# Display the image and the heatmap
def display_gradcam(img_path, heatmap, alpha=0.4):
    img = image.load_img(img_path)
    img = image.img_to_array(img)

    heatmap = cv2.resize(heatmap, (img.shape[1], img.shape[0]))
    heatmap = np.uint8(255 * heatmap)
    heatmap = cv2.applyColorMap(heatmap, cv2.COLORMAP_JET)

    superimposed_img = heatmap * alpha + img
    superimposed_img = np.uint8(superimposed_img)

    plt.imshow(superimposed_img)
    plt.axis('off')
    plt.show()

display_gradcam(img_path, heatmap)
```
